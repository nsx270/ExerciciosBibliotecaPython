2. Gere uma matriz 4x4 com números inteiros de 1 a 16 e extraia a última
linha.

```python
import numpy as np
matriz = np.random.randint(1, 16, (4, 4));
print(matriz);
print("Linha extraída: ")
print(matriz[-1]);
```

3. Crie um array de zeros com comprimento 15 e substitua os valores da 5ª
à 10ª posição por 1.

```python
import numpy as np
array = np.zeros(15);
array[4:10] = 1;
print(array);
```

4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e
a moda dos valores.

```python
import numpy as np
array = np.random.randint(1, 30, size=100);
mediana = np.median(array);
moda = np.bincount(array).argmax();
print("Mediana: ", mediana);
print("Moda: ", moda);
```

5. Crie um array de 20 elementos contendo números entre 1 e 100 e substitua
todos os valores múltiplos de 3 por -3.

```python
import numpy as np
array = np.random.randint(1, 100, size=20);
print("Array original: ",array);
array[array % 3 == 0] = -3;
print("Array modificado: ",array)
```

6. Gere uma matriz 5x5 de números aleatórios entre 0 e 1 e encontre a soma
de cada coluna

```python
import numpy as np
matriz = np.random.rand(5 ,5)
print(matriz)
soma = np.sum(matriz, axis=0)
print(soma)
```

7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

```python
import numpy as np
array1 = np.random.rand(10)
array2 = np.random.rand(10)
correlacao_mt = np.corrcoef(array1, array2)
correlacao = correlacao_mt[0,1]
print("Matriz de correlação: \n", correlacao_mt)
print("Correlação específica: \n", correlacao)
```
