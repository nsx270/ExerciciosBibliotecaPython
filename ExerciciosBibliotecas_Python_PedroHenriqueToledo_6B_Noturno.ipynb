19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np
v1 = np.random.randint(1,10,size=20)
v2 = np.random.randint(1,10,size=20)
cos_theta = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
angulo = np.arccos(np.clip(cos_theta, -1.0, 1.0))
print("Angulo em radianos: ",angulo)

#Cálculo da norma com linalg.norm e produto escalar com dot, aplicando arccos para achar o ângulo em radianos
```

10. Calcule a integral dupla da função f(x,y) = xy + x^2 no intervalo [0,2] para x e [0,1] para y.

```python
from scipy import integrate
def f10(y, x):
  return x * y + x**2
res10, _ = integrate.dblquad(f10, 0, 2, lambda x: 0, lambda x: 1)
print("Integral dupla:", res10)

# dblquad exige a função recebendo (y, x) e os limites de y sendo funções
```

20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x^2.

```python
from matplotlib.ticker import MultipleLocator
x20 = np.linspace(-5, 5, 100)
fig, ax = plt.subplots()
ax.plot(x20, x20**2)
ax.xaxis.set_major_locator(MultipleLocator(1))
ax.yaxis.set_major_locator(MultipleLocator(5))
plt.grid(True)
plt.show()

# Uso de MultipleLocator para ajustar de quanto em quanto a grade aparece
```
